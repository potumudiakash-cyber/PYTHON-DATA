```python
from ast import FunctionType
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import csv
from google.colab import files
uploaded = files.upload()

# Note: Make sure 'filename' is defined before running this cell
x=[]
y=[]
with open('price.csv','r') as csvfile:
    plots=csv.reader(csvfile,delimiter=',')

    next(plots)
    for row in plots:
        x.append((row[0]))
        y.append(float(row[2]))
print(x,y)

plt.plot(x, y, color='blue',
         marker='*',
         linestyle='-',
         ms=8, mfc='red',
         label='CLOSING PRICE')
plt.xticks(rotation=45,fontsize=6)
plt.xlabel('DATES')
plt.ylabel('CLOSING PRICE')
plt.title('STOCK MARKET TREND ANALYSIS',fontsize=6)
plt.grid()
plt.legend()
plt.show()
```

```python
from google.colab import files
uploaded = files.upload()
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
sns.set(style="whitegrid")
dataset=pd.read_csv("vehicle.csv")
dataset.info()
sns.relplot(x="Engine Capacity",
           y="Fuel Efficiency(km/l)",
           data=dataset,
           hue="Vehicle Type",
           s=100,
           marker="*")
plt.xlabel("Engine Capacity")
plt.ylabel("Full Efficiency (km/l)")
plt.title("Vehicle Engine Capacity vs Full Efficiency")
plt.show()
```

```python
from google.colab import files

uploaded = files.upload()
import matplotlib.pyplot as plt
import pandas as pd
data=pd.read_csv("patient.csv")
plt.figure(figsize=(14, 5))
plt.subplot(1,2,1)
sns.lineplot(data=data,x='Month',
             y='Number of patients',
             marker='D',
             color='blue',
             label='Monthly Patient Admission')
plt.xlabel("Month")
plt.ylabel("Number of Patients")
plt.title("Monthly Patient Admission")
plt.legend()
plt.legend()
plt.show()

sns.barplot(data=data,
            x='Month',
            y='Number of Cases',
            color='orange',label='Monthly Emergency Cases')
plt.xlabel("Month")
plt.ylabel("Number of Cases")
plt.title("Monthly Emergency Cases")
plt.tight_layout()
plt.legend()
plt.show()
```
